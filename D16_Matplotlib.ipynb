```python
import matplotlib.pyplot as plt
```

```python
x=["aghan","brian","cathy","david","emily"]
y=[5, 7, 3, 8, 6]

plt.bar(x,y)
plt.xlabel("Countries")
plt.ylabel("Values")
plt.title("Bar Chart")
plt.grid()
plt.show()
```

```python
std_name=["kamran","akmal","sara","ali","zara"]
std_marks=[90,80,33,65,45]

plt.xlabel("Students Names")
plt.ylabel("Marks")
plt.title("Student Report", color="blue")
plt.plot(std_name,std_marks)
plt.bar(std_name,std_marks)
plt.scatter(std_name,std_marks)
```

```python
subjects = ["Math","CS","English"]
marks = [90,85,75]

plt.bar(subjects, marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Report")
plt.show()
```

```python
marks = [45,50,60,70,70,80,85,90,90,95]

plt.hist(marks, bins=8)
plt.show()
```

```python
height = [150,155,160,165]
weight = [50,55,60,65]

plt.scatter( weight, height)
plt.xlabel("Weight")
plt.ylabel("Height")
plt.title("Height vs Weight")

plt.show()
```

```python
sizes = [40,30,20,10]
labels = ["A","B","C","D"]

plt.pie(sizes, labels=labels)
plt.show()

# Displaying percentages:
plt.pie(sizes,
        labels=labels,
        autopct="%1.1f%%")
```
